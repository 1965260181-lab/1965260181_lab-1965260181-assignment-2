```python
import time
from functools import lru_cache


def normalise_and_format(raw_plate: str) -> tuple[bool, str]:

    reg = raw_plate.replace(" ", "").replace("-", "").strip().upper()

    if (
        len(reg) in (9, 10)
        and reg[:2].isalpha()
        and reg[2:4].isdigit()
        and reg[4:-4].isalpha()
        and reg[-4:].isdigit()
    ):
        return True, f"{reg[:2]} {reg[2:4]} {reg[4:-4]} {reg[-4:]}"

    return False, "INVALID"


def record_session_item(
    item: str,
    session_log: list[str] | None = None
) -> list[str]:

    if session_log is None:
        session_log = []

    ok, formatted = normalise_and_format(item)

    if ok:
        session_log.append(formatted)

    return session_log


def build_batch_report(records: list[str]) -> str:

    return "\n".join(records)


def climb_steps_optimal(n: int) -> int:

    if n <= 2:
        return n

    prev2, prev1 = 1, 2

    for _ in range(3, n + 1):
        prev2, prev1 = prev1, prev2 + prev1

    return prev1


if __name__ == "__main__":

    s_a = record_session_item(" tn-09 ab 1234 ")
    s_b = record_session_item("KA-01-M-0007")

    print("Session A:", s_a, "| Session B:", s_b)

    print(
        "Climb (500) computed safely in O(1) space:",
        len(str(climb_steps_optimal(500))),
        "digits"
    )
```
