```python
def subsidy_separate_ifs(units: int, senior_over_80: bool, has_ration_card: bool) -> list[str]:
    awards = []

    if units <= 100:
        awards.append("Eligible (<=100 units)")

    if 101 <= units <= 200 and has_ration_card:
        awards.append("Partial (Ration Card)")

    if units > 200 and senior_over_80:
        awards.append("Eligible (Senior > 80)")

    return awards if awards else ["No Subsidy"]


def subsidy_if_elif_else(units: int, senior_over_80: bool, has_ration_card: bool) -> str:
    if units <= 100 or senior_over_80:
        return "Eligible"
    elif 101 <= units <= 200 and has_ration_card:
        return "Partial Subsidy"
    else:
        return "No Subsidy"


if __name__ == "__main__":
    test_households = [
        (50, False, False),
        (100, False, False),
        (101, False, True),
        (150, False, False),
        (200, False, True),
        (201, False, True),
        (201, True, False),
        (95, True, True)
    ]

    for u, s, r in test_households:
        print(
            f"Units={u:3} | Senior={str(s):5} | "
            f"Ration={str(r):5} -> Separate="
            f"{subsidy_separate_ifs(u,s,r)} | "
            f"Elif={subsidy_if_elif_else(u,s,r)}"
        )
```
