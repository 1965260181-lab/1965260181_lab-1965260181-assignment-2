```python
import time


def build_plus(parts: list[str]) -> str:

    text = ""

    for p in parts:
        text += p + ";"

    return text


def build_join(parts: list[str]) -> str:

    return ";".join(parts) + ";"


if __name__ == "__main__":

    for size in (1_000, 10_000, 200_000):

        items = [f"Item{i}: Rs{i % 50}" for i in range(size)]

        t_p, t_j = 0.0, 0.0

        for _ in range(3):

            s = time.perf_counter()
            a = build_plus(items)
            t_p += time.perf_counter() - s

            s = time.perf_counter()
            b = build_join(items)
            t_j += time.perf_counter() - s

        print(
            f"Size={size:7} | "
            f"Avg +=: {t_p/3:.4f}s | "
            f"Avg join: {t_j/3:.4f}s | "
            f"Match: {a == b}"
        )
```
