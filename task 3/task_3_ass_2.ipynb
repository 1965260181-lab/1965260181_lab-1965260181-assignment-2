```python
def normalise(text: str) -> str:
    return text.replace(" ", "").replace("-", "").strip().upper()


def is_valid(reg: str) -> bool:

    if len(reg) not in (9, 10):
        return False

    state = reg[:2]
    district = reg[2:4]
    series = reg[4:-4]
    number = reg[-4:]

    return (
        state.isalpha()
        and district.isdigit()
        and series.isalpha()
        and number.isdigit()
    )


def format_receipt(reg: str) -> str:

    if not is_valid(reg):
        return "INVALID"

    return f"{reg[:2]} {reg[2:4]} {reg[4:-4]} {reg[-4:]}"


if __name__ == "__main__":

    samples = [
        "tn 09 ab 1234",
        "KA-01-M-0007",
        "TN99AB1234",
        "TN9AB1234",
        "TN 22 1234 AB",
        "MH-12-DE-143"
    ]

    for raw in samples:
        reg = normalise(raw)

        print(
            f"{raw:15} -> {reg:11} "
            f"{'valid' if is_valid(reg) else 'INVALID'} | "
            f"Receipt: {format_receipt(reg)}"
        )
```
