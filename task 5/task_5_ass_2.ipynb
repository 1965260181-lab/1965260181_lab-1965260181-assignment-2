```python
TAX_RATE = 0.05


def add_item_buggy(item: str, order=[]):

    order.append(item)

    return order


def add_item(item: str, order=None):

    if order is None:
        order = []

    order.append(item)

    return order


def calculate_bill(prices: list[float],
                   gst_rate: float = TAX_RATE) -> tuple[float, float, float]:

    subtotal = sum(prices)

    gst = round(subtotal * gst_rate, 2)

    return subtotal, gst, round(subtotal + gst, 2)


if __name__ == "__main__":

    print("Buggy:", add_item_buggy("Tea"), add_item_buggy("Vada"))

    print("Fixed:", add_item("Tea"), add_item("Vada"))

    print("Buggy __defaults__:", add_item_buggy.__defaults__)

    print("Fixed __defaults__:", add_item.__defaults__)

    subtotal, gst, total = calculate_bill([40, 60])

    print(
        f"Billing Helper: Subtotal=Rs.{subtotal:.2f}, "
        f"GST(5%)=Rs.{gst:.2f}, Total=Rs.{total:.2f}"
    )
```
