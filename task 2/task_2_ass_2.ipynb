```python
def verify_otp_while(inputs: list[str], correct_otp: str = "482910",
                     security_passed: bool = False) -> str:

    attempts = 0
    max_attempts = 3
    idx = 0

    while attempts < max_attempts and idx < len(inputs):
        entry = inputs[idx].strip()
        idx += 1

        if entry == "":
            continue

        if entry == correct_otp:
            print("OTP Verified! Access Granted.")
            break

        attempts += 1

        if attempts == 2 and security_passed:
            max_attempts += 2
            security_passed = False

    else:
        if attempts >= max_attempts:
            return "Account locked for 30 minutes."

    return "Session Active"


def verify_otp_for(inputs: list[str], correct_otp: str = "482910") -> str:

    non_blank = [x.strip() for x in inputs if x.strip() != ""]

    for attempt_no in range(3):
        if attempt_no < len(non_blank) and non_blank[attempt_no] == correct_otp:
            return "OTP Verified! Access Granted."
    else:
        return "Account locked for 30 minutes."
        print("While Loop:")
print(verify_otp_while(["123456", "482910"]))

print("\nFor Loop:")
print(verify_otp_for(["123456", "482910"]))
```
