```python
import sys
from functools import lru_cache

plain_calls = 0
memo_calls = 0


def climb_plain(n: int) -> int:

    global plain_calls

    plain_calls += 1

    if n <= 2:
        return n

    return climb_plain(n - 1) + climb_plain(n - 2)


@lru_cache(maxsize=None)
def climb_memo(n: int) -> int:

    global memo_calls

    memo_calls += 1

    if n <= 2:
        return n

    return climb_memo(n - 1) + climb_memo(n - 2)


def climb_iterative(n: int) -> int:

    if n <= 2:
        return n

    a, b = 1, 2

    for _ in range(3, n + 1):
        a, b = b, a + b

    return b


if __name__ == "__main__":

    print(
        "Manual vs Programmatic (n=1..6):",
        [climb_iterative(i) for i in range(1, 7)]
    )

    for n in (10, 20, 25, 30):

        plain_calls = memo_calls = 0

        climb_memo.cache_clear()

        climb_plain(n)
        climb_memo(n)

        print(
            f"n={n:2} | Plain Calls={plain_calls:9,} "
            f"| Memo Calls={memo_calls:3}"
        )
```
